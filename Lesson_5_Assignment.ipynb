```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Task 1: Load the Titanic dataset
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
#Task 2: Inspect the data
# Display the first 5 rows
df.head()
```

```python
#display the last 5 rows
df.tail()
```

```python
#Display the data type
df.dtypes
```

```python
#Display the statistical description
df.describe()
```

```python
#Task 3: Clean and preprocess the data
#Create a new categorical feature 'PclassCategory' from 'Pclass'
df["Pclass_Category"] = df["Pclass"].map({
    1: "First Class",
    2: "Second Class",
    3: "Third Class"
})

df[["Pclass", "Pclass_Category"]].head()
```

```python
#Create a new categorical feature 'CabinCategory' from 'Cabin'
df["Cabin_Category"] = df["Cabin"].str[0]

df[["Cabin", "Cabin_Category"]].head(10)
```

```python
df["Cabin_Category"] = df["Cabin_Category"].fillna("Unknown")

df["Cabin_Category"].value_counts()
```

```python
#Check overall missing values
df.isnull().sum()
```

```python
#Handle missing values in the 'Age' column (use mean age)
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)
print("Missing Age values after cleaning:", df["Age"].isnull().sum())
```

```python
#rerun the code for overall missing values
df.isnull().sum()
```

```python
#Create a 'Family' feature by combining 'SibSp' and 'Parch'
df["Family"] = df["SibSp"] + df["Parch"] + 1
df[["Name", "SibSp", "Parch", "Family"]].head()
```

```python
#Task 4: Identify specific data points
#Find passengers with the highest fare
highest_fare = df["Fare"].max()
df[df["Fare"] == highest_fare]
```

```python
#Passenger(s) with the largest family size
largest_family = df["Family"].max()
print("Largest family size:", largest_family)
df[df["Family"] == largest_family]
```

```python
# Task 5: Demonstrate Pandas Series and DataFrame functionalities
#Create a Pandas Series of passenger names
name_series = pd.Series(["Sage", "Stella", "Bullten", "Federick", "Gladys"])

print(name_series)
```

```python
#Create a DataFrame of passenger names and ages
names_ages_df = pd.Series(["Sage", "Stella", "Bullten", "Federick", "Gladys"])
ages = pd.Series([22, 25, 30, 35, 40])
print(ages)
print(names_ages_df)
```

```python
#Slice the DataFrame to show only passengers older than 50
df[df["Age"] > 50]
```

```python
# Task 6: Introduce multi-indexing
#Create a multi-index DataFrame grouped by 'Sex' and 'Pclass'
multi_index_df = df.groupby(["Sex", "Pclass"])["Survived"].mean().to_frame("Survival_Rate")
multi_index_df
```

```python
# Task 7: Visualizations
# Create a histogram of passenger ages
plt.figure(figsize=(10, 6))
plt.hist(df['Age'], bins=20)
plt.title('Histogram of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
#Create a bar plot of survival rates by passenger class
#group the pessenges per the various classes first before isualizing the bar chart for each class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
plt.figure(figsize=(10, 6))
sns.barplot(
    x=survival_by_class.index,
    y=survival_by_class.values
)
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')

plt.show()
```

```python
#Create a heatmap of correlation between numerical features
plt.figure(figsize=(12, 10))
correlation = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```

```python
#Create a pair plot of 'Age', 'Fare', 'SibSp', and 'Parch', colored by 'Survived'
sns.pairplot(
    df[['Age', 'Fare', 'SibSp', 'Parch', 'Survived']],
    hue='Survived'
)

plt.show()
```

```python
# Task 8: Analyze survival rates
survival_by_sex = df.groupby('Sex')['Survived'].mean()
survival_by_sex
```

```python
#Calculate and display survival rates by passenger class
survival_by_class = df.groupby('Pclass')['Survived'].mean()
survival_by_class
```

```python
# Task 9: Analyze age distribution
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x='Pclass',
    y='Age',
    hue='Sex'
)

plt.title('Age Distribution by Passenger Class and Sex')
plt.xlabel('Passenger Class')
plt.ylabel('Age')

plt.show()
```

```python
# Task 10: Analyze fare distribution
# Create a violin plot of fare distribution by passenger class and survival status
plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x='Pclass',
    y='Fare',
    hue='Survived'
)

plt.title('Fare Distribution by Passenger Class and Survival Status')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')

plt.show()

print("EDA tasks completed.")
```
