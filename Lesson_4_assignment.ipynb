```python
# Task 9: Iterators and Generators
# Create a list of the first 5 even numbers
even_numbers = [0, 2, 4, 6, 8]

# Create an iterator
even_iterator = iter(even_numbers)

# Display the numbers using the iterator
print("Using iterator:")

for number in even_iterator:
    print(number)
```

```python
# Create a generator for the first 5 even numbers
def even_generator():
    for number in range(0, 10, 2):
        yield number

# Use the generator
print("Using generator:")

for number in even_generator():
    print(number)
```

```python
# Task 10: Map, Reduce, and Filter
numbers = [1, 2, 3, 4, 5]

# Use map to square all numbers
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Use reduce to find the product of all numbers
product = reduce(lambda x, y: x * y, numbers)

# Use filter to get only even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

# Display the results
print("Squared numbers:", squared_numbers)
print("Product of all numbers:", product)
print("Even numbers:", even_numbers)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
class Rectangle:
    # Constructor
    def __init__(self, length, width):
        self.length = length
        self.width = width

    # Method to calculate area
    def area(self):
        return self.length * self.width

    # Method to calculate perimeter
    def perimeter(self):
        return 2 * (self.length + self.width)


# Create a Rectangle object
rectangle = Rectangle(10, 5)

# Display the results
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

```python
# Task 12: Pandas Data Analysis
# create teh dataframe for employees and salary first

df_employees = pd.DataFrame({
    'Name': ['Zeenap', 'Paul', 'Sunita', 'Edward', 'Stepene'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})
# 1. Find the average salary by department
average_salary = df_employees.groupby("Department")["Salary"].mean()

print("Average salary by department:")
print(average_salary)
```
